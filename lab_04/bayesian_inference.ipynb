# Задача 1

Пусть $X_1, X_2, \ldots, X_n$ — выборка из экспоненциального распределения с параметром $\lambda$. Найти оценку максимального правдоподобия параметра $\lambda$, сравнить ее с байесовской оценкой (MAP и математическое ожидание апостреорного распределения), подобрав сопряженное распределение. Сравнить полученные байесовские оценки с оценкой MLE. Найти предсказательное распределение

**Решение:**

1. Найти оценку максимального правдоподобия параметра $\lambda$

Функция плотности вероятности:
$$
f(x \mid \lambda) = \lambda e^{-\lambda x}, \quad x \ge 0, \; \lambda > 0
$$

Так как выборка независима, функция правдоподобия выглядит:
$$
L(\lambda) = \prod_{i=1}^n \lambda e^{-\lambda X_i} = \lambda^n e^{-\lambda \sum_{i=1}^n X_i}
$$

Для удобства вычислений "навесим" логарифм:
$$
\ell(\lambda) = \ln L(\lambda) = n \ln \lambda - \lambda \sum_{i=1}^n X_i
$$

Дифференцируем по $\lambda$:
$$
\frac{d\ell}{d\lambda} = \frac{n}{\lambda} - \sum_{i=1}^n X_i
$$
$$
\frac{n}{\lambda} - \sum_{i=1}^n X_i = 0 \quad \Rightarrow \quad \frac{n}{\lambda} = \sum_{i=1}^n X_i
$$
$$
\lambda = \frac{n}{\sum_{i=1}^n X_i}
$$

Проверим на максимум:
$$
\frac{d^2 \ell}{d\lambda^2} = -\frac{n}{\lambda^2} < 0 \quad (\lambda > 0),
$$
значит, найденная точка - точка максимума

$$
\hat{\lambda}_{MLE} = \frac{1}{\overline{X}}
$$

2. Байесовская оценка (MAP и математическое ожидание апостреорного распределения)

Для экспоненциального распределения с неизвестным параметром $\lambda$ сопряжённым априорным распределением является **Гамма-распределение**:
$$
\lambda \sim \text{Gamma}(\alpha, \beta),
$$
с плотностью:

$$
p(\lambda) = \frac{\beta^{\alpha}}{\Gamma(\alpha)} \lambda^{\alpha - 1} e^{-\beta \lambda}, \quad \lambda>0, \alpha>0, \beta>0
$$

Правдоподобие:
$$
p(\mathbf{X}|\lambda) = \prod_{i = 1}^{n} \lambda e^{-\lambda X_i} = \lambda^n e^{-\lambda n \bar{X}}
$$

Апостериорное распределение:
$$
p(\lambda|\mathbf{X}) \propto p(\mathbf{X}|\lambda) \cdot p(\lambda)
$$
$$
\propto \left[ \lambda^n e^{-\lambda n \bar{X}} \right] \cdot \left[ \lambda^{\alpha - 1} e^{-\beta \lambda} \right]
$$
$$
= \lambda^{(\alpha + n) - 1} e^{-(\beta + n\bar{X}) \lambda}
$$

Это ядро $\text{Gamma}(\alpha_n, \beta_n)$, где
$$
\alpha_n = \alpha + n, \quad \beta_n = \beta + n\bar{X}
$$

Таким образом получаем:
$$
\lambda|\mathbf{X} \sim \text{Gamma}(\alpha_n, \beta_n)
$$

**Мода апостериорного:**

Для $\text{Gamma}(\alpha, \beta)$ мода при $\alpha \ge 1$ равна $\frac{\alpha-1}{\beta}$

Если $\alpha_0 + n \ge 1$ (обычно так, т.к. $\alpha_0 > 0$ и $n\ge 1$):

$$
\hat{\lambda}_{\text{MAP}} = \frac{\alpha_n - 1}{\beta_n} = \frac{\alpha + n - 1}{\beta + n\bar{X}}
$$

**Мат. ожидание апостериорного:**

Для $\text{Gamma}(\alpha, \beta):\mathbb{E}[\lambda] = \frac{\alpha}{\beta}$

$$
\hat{\lambda}_{\text{post\_mean}} = \frac{\alpha_n}{\beta_n} = \frac{\alpha + n}{\beta + n\bar{X}}
$$

3. Сравнить полученные байесовские оценки с оценкой MLE
$$
\hat{\lambda}_{MLE} = \frac{1}{\overline{X}}
$$

$$
\hat{\lambda}_{MAP} = \frac{\alpha + n - 1}{\beta + n\bar{X}}
$$

$$
\hat{\lambda}_{\text{post\_mean}} = \frac{\alpha_n}{\beta_n} = \frac{\alpha + n}{\beta + n\bar{X}}
$$

Выводы:
- При $\alpha = 1, \beta = 0$ (неинформативный априор) MAP совпадает с MLE
- При больших n влияние априорных параметров $\alpha$ и $\beta$ становится малым, и все три оценки сходятся к $\frac{1}{\overline{X}}$

4. Найти предсказательное распределение

$$
p(X_{n + 1}|X_n) = \int_{0}^{\infty}p(X_{n + 1}|\lambda)p(\lambda|X)d\lambda =
$$
$$
= \int_0^\infty \left[ \lambda e^{-\lambda X_{n+1}} \right] \cdot \left[ \frac{\beta_n^{\alpha_n}}{\Gamma(\alpha_n)} \lambda^{\alpha_n - 1} e^{-\beta_n \lambda} \right] d\lambda
$$
$$
= \frac{\beta_n^{\alpha_n}}{\Gamma(\alpha_n)} \int_0^\infty \lambda^{\alpha_n} e^{-(\beta_n + X_{n+1})\lambda} \, d\lambda
$$

Интеграл равен $\frac{\Gamma(\alpha_n + 1)}{(\beta_n + X_{n+1})^{\alpha_n+1}}$

Используем $\Gamma(\alpha_n+1) = \alpha_n \Gamma(\alpha_n)$:

$$
p(X_{n+1} | \mathbf{X}) = \frac{\beta_n^{\alpha_n}}{\Gamma(\alpha_n)} \cdot \frac{\Gamma(\alpha_n+1)}{(\beta_n + X_{n+1})^{\alpha_n+1}}
$$
$$
= \frac{\alpha_n \beta_n^{\alpha_n}}{(\beta_n + X_{n+1})^{\alpha_n+1}}, \quad X_{n+1} > 0
$$

Ответ:
$$
\hat{\lambda}_{MLE} = \frac{1}{\overline{X}}
$$

$$
\hat{\lambda}_{MAP} = \frac{\alpha + n - 1}{\beta + n\bar{X}}
$$

$$
\hat{\lambda}_{\text{post\_mean}} = \frac{\alpha_n}{\beta_n} = \frac{\alpha + n}{\beta + n\bar{X}}
$$
$$
p(X_{n+1} | \mathbf{X}) = \frac{\alpha_n \beta_n^{\alpha_n}}{(\beta_n + X_{n+1})^{\alpha_n+1}}
$$

# Задача 2

**Мультиномиальное распределение**

Пусть проводится серия из $n$ испытаний и в результате каждого испытания происходит ровно одно событие из набора $A_1, A_2, \dots, A_m$, причем вероятности этих событий равны соответственно $\mathsf{p}_1, \mathsf{p}_2, \dots, \mathsf{p}_m$, причем
$$\sum_{i=1}^{m}\mathsf{p}_i = 1.$$

Тогда совместное распределение величин $X_1, X_2, \dots, X_m$, где $X_k$ — число наступлений события $A_k$ в серии из $n$ испытаний, задается вероятностями

$$
\mathsf{P}\left(X_1 = n_1, \dots, X_m = n_m, \right) = \frac{n!}{n_1!\dots n_m!}\mathsf{p}_1^{n_1}\dots \mathsf{p}_m^{n_m},
$$

где $n_1, n_2, \dots, n_m$ — произвольный набор целых неотрицательных чисел, таких что

$$\sum_{i=1}^m n_i = n.$$

Произведите байесовский вывод для мультиномиального распределения: найдите апостериорное распределение, используя в качестве сопоряженного распределения к правдоподобию [распределение Дирихле](https://ru.wikipedia.org/wiki/%D0%A0%D0%B0%D1%81%D0%BF%D1%80%D0%B5%D0%B4%D0%B5%D0%BB%D0%B5%D0%BD%D0%B8%D0%B5_%D0%94%D0%B8%D1%80%D0%B8%D1%85%D0%BB%D0%B5), найдите предсказательное распределение. Объясните результат.

**Решение:**

1. Найти апостериорное распределение, используя в качестве сопоряженного распределения к правдоподобию распределение Дирихле

Пусть $\mathbf{X} = (X_1, \dots, X_m)$ — вектор количества наступлений событий $A_1, \dots, A_m$ в $n$ независимых испытаниях,  
где вероятности $\mathbf{p} = (p_1, \dots, p_m)$, $\sum p_i = 1$.

Правдоподобие (мультиномиальное):
$$
L(\mathbf{X}|\mathbf{p}) = \frac{n!}{X_1! \cdots X_m!} p_1^{X_1} \cdots p_m^{X_m},
$$
где $\sum_{i=1}^m X_i = n$.

Множитель $\frac{n!}{X_1! \cdots X_m!}$ не зависит от $\mathbf{p}$, поэтому ядро правдоподобия:
$$
L(\mathbf{X}|\mathbf{p}) \propto p_1^{X_1} \cdots p_m^{X_m}
$$

Плотность распределения Дирихле с параметрами $\boldsymbol{\alpha} = (\alpha_1, \dots, \alpha_m)$, $\alpha_i > 0$:
$$
p(\mathbf{p} \mid \boldsymbol{\alpha}) = \frac{\Gamma\left( \sum_{i=1}^m \alpha_i \right)}{\prod_{i=1}^m \Gamma(\alpha_i)} \prod_{i=1}^m p_i^{\alpha_i - 1},
$$
где $\sum p_i = 1$, $p_i \ge 0$.

Среднее априорное:  
$$
\mathbb{E}[p_i] = \frac{\alpha_i}{\alpha_0},
$$
где $\alpha_0 = \sum_{i=1}^m \alpha_i$.

Модой распределения является вектор $x = (x_1, \dots, x_m)$ c
$$
x_i = \frac{\alpha_i - 1}{\alpha_0 - 1}, \quad \alpha_i > 1
$$


По теореме Байеса:
$$
p(\mathbf{p} \mid \mathbf{X}) \propto p(\mathbf{X}|\mathbf{p}) \, p(\mathbf{p} \mid \boldsymbol{\alpha})
$$

Подставляем:
$$
p(\mathbf{p} \mid \mathbf{X}) \propto \left[ \prod_{i=1}^m p_i^{X_i} \right] \cdot \left[ \prod_{i=1}^m p_i^{\alpha_i - 1} \right]
= \prod_{i=1}^m p_i^{\alpha_i + X_i - 1}
$$

Это ядро распределения Дирихле с обновлёнными параметрами:
$$
\boldsymbol{\alpha}_{\text{post}} = (\alpha_1 + X_1, \dots, \alpha_m + X_m)
$$

Таким образом:
$$
\mathbf{p} \mid \mathbf{X} \sim \text{Dirichlet}(\alpha_1 + X_1, \dots, \alpha_m + X_m)
$$

**Апостериорное среднее:**
$$
\mathbb{E}[p_i \mid \mathbf{X}] = \frac{\alpha_i + X_i}{\sum_{j=1}^m (\alpha_j + X_j)} = \frac{\alpha_i + X_i}{\alpha_0 + n},
$$
поскольку $\sum X_j = n$.

**MAP (мода апостериорного):**  

Для Дирихле мода при $\alpha_i > 1$:
$$
\hat{p}_i^{\text{MAP}} = \frac{\alpha_i + X_i - 1}{\sum_{j=1}^m (\alpha_j + X_j - 1)} = \frac{\alpha_i + X_i - 1}{\alpha_0 + n - m}
$$

2. Найти предсказательное распределение
Необходимо найти вероятность $P(X_{n+1} = k \mid \mathbf{X})$ - какая категория появится в следующем испытании.

По правилу усреднения по апостериорному:
$$
P(X_{n+1} = k \mid \mathbf{X}) = \mathbb{E}[p_k \mid \mathbf{X}] = \frac{\alpha_k + X_k}{\alpha_0 + n}
$$

Это и есть **предсказательное распределение** для одного нового наблюдения.

Выводы:
- **Апостериорное среднее** - выпуклая комбинация (сумма коэффициентов равна 1) априорного среднего $\frac{\alpha_i}{\alpha_0}$ и эмпирической частоты $\frac{X_i}{n}$:

$$
\mathbb{E}[p_i \mid \mathbf{X}] = \frac{\alpha_0}{\alpha_0 + n} \cdot \frac{\alpha_i}{\alpha_0} + \frac{n}{\alpha_0 + n} \cdot \frac{X_i}{n}.
$$

- **Псевдо-наблюдения**: параметры $\alpha_i$ интерпретируются как количество "воображаемых" наблюдений категории $i$ до начала эксперимента. После получения данных эти псевдо-наблюдения просто складываются с реальными.

Итог:
**Априор**: 
$$
\mathbf{p} \sim \text{Dirichlet}(\alpha_1, \dots, \alpha_m)
$$

**Апостериор**:
$$
\mathbf{p} \mid \mathbf{X} \sim \text{Dirichlet}(\alpha_1+X_1, \dots, \alpha_m+X_m)
$$

**Апостериорное среднее**:  
$$
\mathbb{E}[p_i \mid \mathbf{X}] = \frac{\alpha_i + X_i}{\alpha_0 + n}
$$

**MAP-оценка**:
$$
\hat{p}_i^{\text{MAP}} = \frac{\alpha_i + X_i - 1}{\sum_{j=1}^m (\alpha_j + X_j - 1)} = \frac{\alpha_i + X_i - 1}{\alpha_0 + n - m}
$$

**Предсказательное распределение** для одного нового испытания:  
$$
P(\text{категория } i \mid \mathbf{X}) = \frac{\alpha_i + X_i}{\alpha_0 + n}
$$
